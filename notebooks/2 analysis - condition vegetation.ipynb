###  Berm Condition & Vegetation Analysis


**Research questions**

1. What is predictive of berm condition / intactness? 
2. What is predictive of berm impact on surrounding vegetation, 
3. What is the correspondence between these condition and vegetation response, if any?

Methods: 
- Variables are highly correlated, we first examined which variables explain highest fraction variance in SAVI 


Results:
 - Longer berms more likely to be degraded. Perhaps because more failure points 
 - Landform is not the best predictor, but useful as a conceptual framework – correspondence to berm purpose (are some water spreader berms actually flood control berms in the dataset).

-  Fan terraces are more effective because steeper,  possibly more runoff capture
-  Floodplains less effective because gentle slopes, possibly high clay

Different reasons, design criteria, inherent differnces.

Flow accumulation tends to be higher in floodplains --> intact berms tend to have lower flow accumulation

## Figure 2 — Model performance and feature importance

```python
# ── Fig 2: RF permutation importance (both outcomes) ────────────────────
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.inspection import permutation_importance
from matplotlib.lines import Line2D
from pathlib import Path

_FS = FS
_SEED = 42
_N_REPEATS = 30

# ── Prepare data ─────────────────────────────────────────────────────────
_df = data.dropna(subset=["Intact", "Effective"]).copy()

# Encode categorical predictors
_pred_cols = ["slope_200", "FA_30_max", "Shape_Leng", "Landform", "Texture", "Soil_Development"]
_df_model = _df.dropna(subset=_pred_cols).copy()

_X_num = _df_model[["slope_200", "FA_30_max", "Shape_Leng"]]
_X_cat = pd.get_dummies(_df_model[["Landform", "Texture", "Soil_Development"]], drop_first=True)
_X = pd.concat([_X_num, _X_cat], axis=1).fillna(0)

_y_cond = _df_model["Intact"].astype(int)
_y_veg  = _df_model["Effective"].map({"Vegetation response": 1, "No vegetation response": 0}).astype(int)

# ── Fit RF models ────────────────────────────────────────────────────────
_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=_SEED)
_rf_kwargs = dict(n_estimators=300, class_weight="balanced", random_state=_SEED)

# Condition
_rf_cond = RandomForestClassifier(**_rf_kwargs)
_auc_cond = cross_val_score(_rf_cond, _X, _y_cond, cv=_cv, scoring="roc_auc")
_rf_cond.fit(_X, _y_cond)
_perm_cond = permutation_importance(_rf_cond, _X, _y_cond, n_repeats=_N_REPEATS,
                                     scoring="roc_auc", random_state=_SEED)

# Vegetation response
_rf_veg = RandomForestClassifier(**_rf_kwargs)
_auc_veg = cross_val_score(_rf_veg, _X, _y_veg, cv=_cv, scoring="roc_auc")
_rf_veg.fit(_X, _y_veg)
_perm_veg = permutation_importance(_rf_veg, _X, _y_veg, n_repeats=_N_REPEATS,
                                    scoring="roc_auc", random_state=_SEED)

print(f"Condition: CV AUC = {_auc_cond.mean():.3f} \u00b1 {_auc_cond.std():.3f}")
print(f"Vegetation: CV AUC = {_auc_veg.mean():.3f} \u00b1 {_auc_veg.std():.3f}")

# ── Aggregate importances by predictor group ──────────────────────────────
_col_to_group = {}
for c in _X.columns:
    matched = False
    for parent in ["Landform", "Texture", "Soil_Development"]:
        if c.startswith(parent + "_"):
            _col_to_group[c] = parent
            matched = True
            break
    if not matched:
        _col_to_group[c] = c

_groups = list(dict.fromkeys(_col_to_group.values()))

def _aggregate_importance(perm_result, col_to_group, groups, columns):
    """Sum permutation importance across one-hot columns for each parent predictor."""
    imp_mean = {}
    imp_std  = {}
    for g in groups:
        idxs = [i for i, c in enumerate(columns) if col_to_group[c] == g]
        imp_mean[g] = sum(perm_result.importances_mean[i] for i in idxs)
        imp_std[g] = np.sqrt(sum(perm_result.importances_std[i]**2 for i in idxs))
    return imp_mean, imp_std

_imp_cond_m, _imp_cond_s = _aggregate_importance(_perm_cond, _col_to_group, _groups, list(_X.columns))
_imp_veg_m,  _imp_veg_s  = _aggregate_importance(_perm_veg,  _col_to_group, _groups, list(_X.columns))

_sorted_groups = sorted(_groups, key=lambda g: _imp_cond_m[g], reverse=True)

_pretty = {
    "slope_200": "Hillslope gradient",
    "FA_30_max": "Flow accumulation",
    "Shape_Leng": "Berm length",
    "Landform": "Landform",
    "Texture": "Soil texture",
    "Soil_Development": "Soil development",
}

# ── 2-panel figure ───────────────────────────────────────────────────────
_fig2, (_ax_imp_c, _ax_imp_v) = plt.subplots(1, 2, figsize=(14, 6),
    gridspec_kw={"width_ratios": [1, 1]})
_fig2.subplots_adjust(wspace=0.45)

# ── A: Condition importance ──────────────────────────────────────────────
_y_pos = np.arange(len(_sorted_groups))
_vals_c = [_imp_cond_m[g] for g in _sorted_groups]
_errs_c = [_imp_cond_s[g] for g in _sorted_groups]
_labels = [_pretty.get(g, g) for g in _sorted_groups]
_ax_imp_c.barh(_y_pos, _vals_c, xerr=_errs_c, color=MODEL_CLR_CONDITION,
               edgecolor="white", height=0.6, capsize=3, zorder=3)
_ax_imp_c.set_yticks(_y_pos)
_ax_imp_c.set_yticklabels(_labels, fontsize=_FS)
_ax_imp_c.invert_yaxis()
_ax_imp_c.axvline(0, color="black", lw=0.7)
_ax_imp_c.set_xlabel("Permutation importance\n(decrease in AUC)", fontsize=_FS)
_ax_imp_c.set_title(f"A  Structural condition (AUC = {_auc_cond.mean():.2f})", fontsize=_FS, loc="left")
_ax_imp_c.spines[["top", "right"]].set_visible(False)
_ax_imp_c.tick_params(labelsize=_FS)

# ── B: Vegetation importance ────────────────────────────────────────────
_sorted_veg = sorted(_groups, key=lambda g: _imp_veg_m[g], reverse=True)
_vals_v = [_imp_veg_m[g] for g in _sorted_veg]
_errs_v = [_imp_veg_s[g] for g in _sorted_veg]
_labels_v = [_pretty.get(g, g) for g in _sorted_veg]
_ax_imp_v.barh(_y_pos, _vals_v, xerr=_errs_v, color=MODEL_CLR_VEGRESPONSE,
               edgecolor="white", height=0.6, capsize=3, zorder=3)
_ax_imp_v.set_yticks(_y_pos)
_ax_imp_v.set_yticklabels(_labels_v, fontsize=_FS)
_ax_imp_v.invert_yaxis()
_ax_imp_v.axvline(0, color="black", lw=0.7)
_ax_imp_v.set_xlabel("Permutation importance\n(decrease in AUC)", fontsize=_FS)
_ax_imp_v.set_title(f"B  Vegetation response (AUC = {_auc_veg.mean():.2f})", fontsize=_FS, loc="left")
_ax_imp_v.spines[["top", "right"]].set_visible(False)
_ax_imp_v.tick_params(labelsize=_FS)

_fig2.suptitle("Random Forest \u2014 predictor importance",
               fontsize=FS_SUP, y=1.02)
plt.tight_layout()

# ── Save and register ────────────────────────────────────────────────────
_out2 = Path("../figures/outcomes/fig5_model_performance_importance.png")
_out2.parent.mkdir(parents=True, exist_ok=True)
_fig2.savefig(_out2, dpi=300, bbox_inches="tight")

register_outcomes_figure(
    "FIG_5",
    label="fig5",
    file_name=_out2.name,
    stats_text=(
        f"Random Forest (n_estimators=300, class_weight='balanced'), 5-fold stratified CV. "
        f"Permutation importance = mean decrease in holdout AUC from shuffling each predictor "
        f"({_N_REPEATS} repeats). Importances aggregated across one-hot columns for categorical "
        f"predictors. "
        f"CV AUC (condition) = {_auc_cond.mean():.2f} \u00b1 {_auc_cond.std():.2f} (n={len(_y_cond)}). "
        f"CV AUC (vegetation response) = {_auc_veg.mean():.2f} \u00b1 {_auc_veg.std():.2f} (n={len(_y_veg)})."
    ),
    interpretation_text=(
        f"Panel A: the predictor rankings differ between outcomes \u2014 flow accumulation and berm "
        "length dominate condition prediction, while hillslope gradient and soil properties are "
        "more important for vegetation response. "
        f"CV AUC: condition = {_auc_cond.mean():.2f}, vegetation response = {_auc_veg.mean():.2f}."
    ),
)

plt.show()
```

```python

from pathlib import Path
from scipy.stats import chi2_contingency, fisher_exact
from itertools import combinations as _combinations

_w = 0.25   # bar half-width for this cell

# Length labels (matching len_bin values and LENGTH_COLORS keys)
long_lbl  = "Long (> 60 m)"
short_lbl = "Short (≤ 60 m)"

thr = 60   # length threshold in metres (matches Berm_Length_Class)

# Slope labels (matching Slope_Class values created earlier)
shallow_lbl = "Shallow (≤ 2%)"
steep_lbl   = "Steep (> 2%)"

# B horizon labels (matching Soil_Development values)
bhorizon_lbl   = "B horizon"
nobhorizon_lbl = "No B horizon"

# ── Flow accumulation binarization: find threshold that maximises significance ──
_fa_col = "FA_30_max"
_fa_vals = data[_fa_col].dropna().sort_values().unique()
# candidate thresholds: every unique value between 10th and 90th percentile
#   (avoid near-empty groups at the tails)
_p10, _p90 = np.percentile(_fa_vals, 10), np.percentile(_fa_vals, 90)
_candidates = _fa_vals[(_fa_vals > _p10) & (_fa_vals < _p90)]

_best_p, _best_thr = 1.0, _fa_vals[len(_fa_vals)//2]  # fallback = median
_d_fa = data.dropna(subset=[_fa_col, "Effective"]).copy()
for _thr in _candidates:
    _hi = _d_fa[_fa_col] >= _thr
    _lo = ~_hi
    _k_hi = (_d_fa.loc[_hi, "Effective"] == LBL_EFFECTIVE).sum()
    _n_hi = _hi.sum()
    _k_lo = (_d_fa.loc[_lo, "Effective"] == LBL_EFFECTIVE).sum()
    _n_lo = _lo.sum()
    if _n_hi == 0 or _n_lo == 0:
        continue
    _, _p = fisher_exact([[_k_hi, _n_hi - _k_hi], [_k_lo, _n_lo - _k_lo]])
    if _p < _best_p:
        _best_p, _best_thr = _p, _thr

_fa_opt = _best_thr
_fa_opt = 2000 # override with round number for interpretability (and because the exact threshold is likely not meaningful given data limitations)
print(f"Optimal FA threshold: {_fa_opt:,.0f}  (p = {_best_p:.4g})")

data["FA_Class"] = np.where(data[_fa_col] >= _fa_opt,
                            f"High FA (≥ {_fa_opt/1000:.0f}k)",
                            f"Low FA (< {_fa_opt/1000:.0f}k)")
_fa_high_lbl = f"High FA (≥ {_fa_opt/1000:.0f}k)"
_fa_low_lbl  = f"Low FA (< {_fa_opt/1000:.0f}k)"
FA_COLORS = {_fa_high_lbl: "#4a90d9", _fa_low_lbl: "#a8c8f0"}

# ── Helpers (vegetation-response versions) ────────────────────────────────

def _eff_two_cat_metrics(df_sub, group_col, cat_a, cat_b):
    """Fisher exact for vegetation response rate between two categories."""
    d = df_sub.dropna(subset=[group_col, "Effective"]).copy()
    n_a = (d[group_col] == cat_a).sum()
    n_b = (d[group_col] == cat_b).sum()
    rows = {}
    for metric, pos_fn in [(LBL_EFFECTIVE,   lambda x: x["Effective"] == LBL_EFFECTIVE),
                            (LBL_INEFFECTIVE, lambda x: x["Effective"] == LBL_INEFFECTIVE)]:
        mask = pos_fn(d)
        c_a  = mask[d[group_col] == cat_a].sum()
        c_b  = mask[d[group_col] == cat_b].sum()
        p_a, p_b = c_a / n_a, c_b / n_b
        _, p_val = fisher_exact([[c_a, n_a - c_a], [c_b, n_b - c_b]])
        rows[metric] = {cat_a: p_a, cat_b: p_b, "sig": _sig_stars(p_val)}
    return pd.DataFrame(rows).T

def _draw_eff_two_cat_panel(ax, m, cat_a, cat_b, colors, legend_title, title):
    outcomes = [LBL_EFFECTIVE, LBL_INEFFECTIVE]
    short    = ["Vegetation response", "No vegetation response"]
    x = np.arange(2)
    for cat, offset in [(cat_a, -_w/2), (cat_b, _w/2)]:
        vals = [m.loc[o, cat] for o in outcomes]
        ax.bar(x + offset, vals, _w, label=cat, color=colors[cat],
               edgecolor="black", linewidth=0.7)
    for i, o in enumerate(outcomes):
        sig = m.loc[o, "sig"]
        top = max(m.loc[o, cat_a], m.loc[o, cat_b]) + 0.03
        bh  = 0.015
        ax.plot([x[i]-_w/2, x[i]-_w/2, x[i]+_w/2, x[i]+_w/2],
                [top, top+bh, top+bh, top], lw=1, color="black")
        col = "black" if sig != "ns" else "#888888"
        ax.text(x[i], top+bh+0.005, sig, ha="center", va="bottom",
                fontsize=FS, color=col, fontweight="bold" if sig != "ns" else "normal")
    ax.set_xticks(x)
    ax.set_xticklabels(short)
    ax.set_ylabel("Proportion of berms")
    
    ax.set_title(f"{title}", fontsize=FS, loc="left", x = -0.05, y = 1.05)
    
    ax.set_ylim(0, 1.05)
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    
    ax.legend(frameon=True, fontsize=FS, loc="upper right",bbox_to_anchor=(1, 1.05))
    sns.despine(ax=ax)

def _draw_eff_multi_cat_panel(ax, df_sub, group_col, cat_order, cat_colors, title, pairwise=False):
    """Grouped bars: Response / No response x-axis, one bar per category."""
    d = df_sub.dropna(subset=[group_col, "Effective"]).copy()
    cats   = [c for c in cat_order if c in d[group_col].unique()]
    n_cats = len(cats)
    bar_w  = _w
    _mc_w  = n_cats * bar_w
    offsets = np.linspace(-_mc_w/2 + bar_w/2, _mc_w/2 - bar_w/2, n_cats)
    outcomes = [LBL_EFFECTIVE, LBL_INEFFECTIVE]
    short    = ["Vegetation response", "No vegetation response"]
    x = np.arange(2)

    bars_data = {}
    ns_cat = {}
    for cat in cats:
        sub  = d[d[group_col] == cat]
        n    = len(sub)
        ns_cat[cat] = n
        bars_data[cat] = {
            LBL_EFFECTIVE:   (sub["Effective"] == LBL_EFFECTIVE).sum() / n,
            LBL_INEFFECTIVE: (sub["Effective"] == LBL_INEFFECTIVE).sum() / n,
        }

    # overall chi-sq for title
    ct_chi = pd.crosstab(d[group_col], d["Effective"]).reindex(cats).dropna()
    _, p_chi, *_ = chi2_contingency(ct_chi)
    p_str = f"p = {p_chi:.3g}" if p_chi >= 0.001 else "p < 0.001"

    for ci, (cat, offset) in enumerate(zip(cats, offsets)):
        for oi, outcome in enumerate(outcomes):
            ax.bar(x[oi] + offset, bars_data[cat][outcome], bar_w,
                   color=cat_colors.get(cat, "#aaaaaa"), edgecolor="black", linewidth=0.5,
                   label=cat if oi == 0 else "_nolegend_")

    bh = 0.025

    if pairwise:
        pair_list = list(_combinations(range(n_cats), 2))
        pair_list.sort(key=lambda p: abs(offsets[p[1]] - offsets[p[0]]))

        # Pre-pass: compute all raw p-values, then BH-FDR adjust across
        # all (outcome × pair) tests in this panel.
        _raw = {}
        for oi, outcome in enumerate(outcomes):
            for ci, cj in pair_list:
                ca, cb = cats[ci], cats[cj]
                na, nb = ns_cat[ca], ns_cat[cb]
                ka = (d[d[group_col]==ca]["Effective"] == outcome).sum()
                kb = (d[d[group_col]==cb]["Effective"] == outcome).sum()
                _, p_pair = fisher_exact([[ka, na-ka], [kb, nb-kb]])
                _raw[(oi, ci, cj)] = p_pair
        _keys = list(_raw.keys())
        _ps   = np.array([_raw[k] for k in _keys])
        _m    = len(_ps)
        _ord  = np.argsort(_ps)
        _adj  = np.empty(_m)
        _running = 1.0
        for _rank in range(_m - 1, -1, -1):
            _idx = _ord[_rank]
            _val = _ps[_idx] * _m / (_rank + 1)
            _running = min(_running, _val)
            _adj[_idx] = min(_running, 1.0)
        _padj = {k: _adj[i] for i, k in enumerate(_keys)}

        global_y_max = 0
        for oi, outcome in enumerate(outcomes):
            current_top = max(bars_data[cat][outcome] for cat in cats) + 0.03
            for ci, cj in pair_list:
                ca, cb = cats[ci], cats[cj]
                p_pair = _padj[(oi, ci, cj)]
                sig = _sig_stars(p_pair)
                x_left  = x[oi] + offsets[ci]
                x_right = x[oi] + offsets[cj]
                y_base  = max(bars_data[ca][outcome], bars_data[cb][outcome])
                y_start = max(y_base + 0.02, current_top)
                ax.plot([x_left, x_left, x_right, x_right],
                        [y_start, y_start+bh, y_start+bh, y_start],
                        lw=0.9, color="black" if sig != "ns" else "#bbbbbb")
                col = "black" if sig != "ns" else "#aaaaaa"
                ax.text((x_left+x_right)/2, y_start+bh+0.003, sig,
                        ha="center", va="bottom", fontsize=FS, color=col,
                        fontweight="bold" if sig != "ns" else "normal")
                current_top = y_start + bh + 0.025
            global_y_max = max(global_y_max, current_top)
        y_max = global_y_max + 0.04
    else:
        sig_overall = _sig_stars(p_chi)
        for oi in range(2):
            vals = [bars_data[cat][outcomes[oi]] for cat in cats]
            top  = max(vals) + 0.03
            ax.plot([x[oi]-_mc_w/2, x[oi]-_mc_w/2, x[oi]+_mc_w/2, x[oi]+_mc_w/2],
                    [top, top+bh, top+bh, top], lw=1, color="black")
            col = "black" if sig_overall != "ns" else "#888888"
            ax.text(x[oi], top+bh+0.005, sig_overall, ha="center", va="bottom",
                    fontsize=FS, color=col, fontweight="bold" if sig_overall != "ns" else "normal")
        y_max = max(max(bars_data[cat][o] for cat in cats) for o in outcomes) + 0.12

    ax.set_xticks(x)
    ax.set_xticklabels(short)
    ax.set_ylabel("Proportion of berms")
    ax.set_title(f"{title}", fontsize=FS, loc="left", x = -0.05, y = 1.05)
    ax.set_ylim(0, 1.1)
    ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    ax.legend(frameon=True, fontsize=FS, loc="upper right",bbox_to_anchor=(1, 1.05))
    sns.despine(ax=ax)
    

# ── Two-category vegetation-response metrics ───────────────────────────────
eff_len_m      = _eff_two_cat_metrics(data, "Berm_Length_Class", long_lbl, short_lbl)
eff_slope_m    = _eff_two_cat_metrics(data, "Slope_Class", shallow_lbl, steep_lbl)
eff_bhorizon_m = _eff_two_cat_metrics(data, "Soil_Development", bhorizon_lbl, nobhorizon_lbl)
eff_fa_m       = _eff_two_cat_metrics(data, "FA_Class", _fa_high_lbl, _fa_low_lbl)

# Use canonical colours from constants (keyed on the canonical 50m labels)
EFF_COLORS_LEN   = {long_lbl:  LENGTH_COLORS["Long (> 60 m)"],
                    short_lbl: LENGTH_COLORS["Short (≤ 60 m)"]}
EFF_COLORS_SLOPE = {shallow_lbl: SLOPE_COLORS["Shallow (≤ 2%)"],
                    steep_lbl:   SLOPE_COLORS["Steep (> 2%)"]}
EFF_COLORS_BHORIZON = {bhorizon_lbl:   SOILDEV_COLORS["B horizon"],
                       nobhorizon_lbl: SOILDEV_COLORS["No B horizon"]}

# ── Texture colours — use canonical palette from constants ──
tex_counts2    = data["Texture"].value_counts()
tex_valid2     = tex_counts2[tex_counts2 > 100].index
tex_order_eff  = [t for t in texture_order if t in tex_valid2]
tex_colors_eff = {t: TEXTURE_COLORS.get(t, "#aaaaaa") for t in tex_order_eff}

# ── 2×3 figure ────────────────────────────────────────────────────────────
# Layout: row 0 = [length, slope, landform]
#         row 1 = [texture, B horizon, FA]
plt.rcParams.update({"font.size": 12, "axes.titlesize": 12, "axes.labelsize": 12,
                     "xtick.labelsize": 12, "ytick.labelsize": 12})
fig, axes = plt.subplots(2, 3, figsize=(20, 11), sharey = True)
plt.subplots_adjust(wspace=0.3, hspace=0.4)
_draw_eff_two_cat_panel(axes[0, 0], eff_len_m, long_lbl, short_lbl,
    EFF_COLORS_LEN, "Length class",
    "A  |  Berm length")

_draw_eff_two_cat_panel(axes[0, 1], eff_slope_m, shallow_lbl, steep_lbl,
    EFF_COLORS_SLOPE, "Slope class",
    "B  |  Slope class")

_draw_eff_multi_cat_panel(axes[0, 2], data, "Landform",
    lf_order, LF_COLORS,
    "C  |  Landform", pairwise=True)

_draw_eff_multi_cat_panel(axes[1, 0], data, "Texture",
    tex_order_eff, tex_colors_eff,
    "D  |  Soil texture", pairwise=True)

_draw_eff_two_cat_panel(axes[1, 1], eff_bhorizon_m, bhorizon_lbl, nobhorizon_lbl,
    EFF_COLORS_BHORIZON, "Soil development",
    "E  |  B horizon presence")

_draw_eff_two_cat_panel(axes[1, 2], eff_fa_m, _fa_high_lbl, _fa_low_lbl,
    FA_COLORS, "Flow accumulation",
    "F  |  Flow accumulation")

# Shared y: drop redundant y labels on non-leftmost columns
for _ax in axes[:, 1:].flat:
    _ax.set_ylabel("")

# plt.suptitle("Vegetation response by predictor class", fontsize=FS_SUP, y=1.01)
# plt.tight_layout()


# ── Save and register as Paper 1 Figure 4 ─────────────────────────────────
_out = Path("../figures/outcomes/fig7_vegetation_response.png")
_out.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(_out, dpi=300, bbox_inches="tight")

register_outcomes_figure(
    "FIG_7",
    label="fig7",
    file_name=_out.name,
    stats_text=(
        "2x3 panel figure. "
        "Row 1: berm length, slope class, landform. "
        "Row 2: soil texture, B horizon presence, flow accumulation. "
        "Two-category panels (length, slope, B horizon, flow accumulation) use two-sided Fisher's exact tests. "
        "Multi-category panels (landform, texture) use pairwise two-sided Fisher's exact tests with "
        "Benjamini-Hochberg FDR correction. "
        f"Flow accumulation (30 m) binarized at {_fa_opt/1000:.0f}k contributing cells "
        f"(two-sided Fisher p = {_best_p:.4g})."
    ),
    interpretation_text=(
        "Six predictors of vegetation response shown in a 2x3 grid. "
        f"Row 1: berm length (threshold {thr} m), slope class, and landform. "
        "Row 2: soil texture (n > 100 per class), B horizon presence, and flow accumulation. "
        "Vegetation response is significantly associated with slope class, landform, and soil texture. "
        "Steep-slope berms and flood-plain berms show the highest rates of vegetation response."
    ),
)

plt.show()
```

tests with Benjamini–Hochberg FDR correction** — i.e. the same comparisons
shown by the brackets in Figs. 7 and 8. The "vs … (p)" column lists, for
each level, the other levels it differs from significantly (FDR-adjusted
p < 0.05). Flow accumulation is binarised at the same 2k contributing-cells
threshold for both outcomes. CSV saved to
`data/summary/univariate_outcomes_table.csv`.

```python
# ── Univariate results table for Figs. 7 & 8 ─────────────────────────────
# One row per (predictor × level). Statistics are matched to the figure
# brackets:
#   • binary predictors: two-sided Fisher's exact (single p-value per row)
#   • multi-category predictors: pairwise two-sided Fisher's exact with
#     Benjamini–Hochberg FDR correction across all pairs within that
#     (predictor × outcome). The "vs ... (p)" column lists the significantly
#     different levels (FDR-adjusted p < 0.05) and the adjusted p-value.

from pathlib import Path
from itertools import combinations as _combinations
from scipy.stats import fisher_exact

def _fmt_p(p):
    if pd.isna(p):
        return ""
    if p < 0.001:
        return "<0.001"
    return f"{p:.3f}"

def _stars(p):
    if pd.isna(p):
        return ""
    if p < 0.001: return "***"
    if p < 0.01:  return "**"
    if p < 0.05:  return "*"
    return ""

def _bh(ps):
    """Benjamini–Hochberg FDR adjustment. ps is a 1-D array."""
    ps = np.asarray(ps, dtype=float)
    m = len(ps)
    order = np.argsort(ps)
    adj = np.empty(m)
    running = 1.0
    for rank in range(m - 1, -1, -1):
        idx = order[rank]
        val = ps[idx] * m / (rank + 1)
        running = min(running, val)
        adj[idx] = min(running, 1.0)
    return adj

def _two_cat_row(df, group_col, cat_a, cat_b, outcome_col, pos_label):
    d = df.dropna(subset=[group_col, outcome_col]).copy()
    sub_a = d[d[group_col] == cat_a]
    sub_b = d[d[group_col] == cat_b]
    n_a, n_b = len(sub_a), len(sub_b)
    k_a = (sub_a[outcome_col] == pos_label).sum()
    k_b = (sub_b[outcome_col] == pos_label).sum()
    p_a = k_a / n_a if n_a else np.nan
    p_b = k_b / n_b if n_b else np.nan
    _, fisher_p = fisher_exact([[k_a, n_a - k_a], [k_b, n_b - k_b]])
    return n_a, p_a, n_b, p_b, fisher_p

def _multi_cat_summary(df, group_col, cat_order, outcome_col, pos_label):
    """For a multi-cat predictor + outcome, compute per-level (n, %positive)
    plus FDR-adjusted pairwise Fisher p-values.

    Returns (rows, padj) where:
      rows = [(level, n, p_pos), ...]
      padj = {(i, j): adj_p}  for i < j into the rows list
    """
    d = df.dropna(subset=[group_col, outcome_col]).copy()
    cats = [c for c in cat_order if c in d[group_col].unique()]
    rows = []
    for c in cats:
        sub = d[d[group_col] == c]
        n   = len(sub)
        p   = (sub[outcome_col] == pos_label).sum() / n if n else np.nan
        rows.append((c, n, p))
    pairs = list(_combinations(range(len(cats)), 2))
    raw_p = []
    for i, j in pairs:
        ka = (d[d[group_col] == cats[i]][outcome_col] == pos_label).sum()
        na = (d[group_col] == cats[i]).sum()
        kb = (d[d[group_col] == cats[j]][outcome_col] == pos_label).sum()
        nb = (d[group_col] == cats[j]).sum()
        _, p = fisher_exact([[ka, na - ka], [kb, nb - kb]])
        raw_p.append(p)
    adj = _bh(raw_p) if raw_p else np.array([])
    padj = {pairs[k]: adj[k] for k in range(len(pairs))}
    return rows, padj

def _sig_pairs_string(level_idx, levels, padj, alpha=0.05):
    """Return e.g. 'vs Flood plains (0.012)*; vs Stream terraces (0.041)*'."""
    parts = []
    for (i, j), p in padj.items():
        if p >= alpha:
            continue
        if i == level_idx:
            other = levels[j]
        elif j == level_idx:
            other = levels[i]
        else:
            continue
        parts.append(f"vs {other} ({p:.3f}){_stars(p)}")
    return "; ".join(parts)

# Working copy
_uni = data.copy()
_uni["Condition"] = np.where(_uni["Intact"].astype(bool), "Intact", "Degraded")

# Texture: limit to classes with n > 100 (matches Fig. 7/8 panels)
_tex_counts = _uni["Texture"].value_counts()
_tex_keep   = _tex_counts[_tex_counts > 100].index.tolist()
_tex_order_tbl = [t for t in texture_order if t in _tex_keep]

# Use the single 2k FA cutoff (FA_Class) for both outcomes.
PRED_SPECS = [
    # (display name, group column, kind, levels)
    ("Berm length",          "Berm_Length_Class", "binary",
        (long_lbl, short_lbl)),
    ("Slope class",          "Slope_Class",       "binary",
        (shallow_lbl, steep_lbl)),
    ("Landform",             "Landform",          "multi", lf_order),
    ("Soil texture (n>100)", "Texture",           "multi", _tex_order_tbl),
    ("B horizon presence",   "Soil_Development",  "binary",
        (bhorizon_lbl, nobhorizon_lbl)),
    ("Flow accumulation",    "FA_Class",          "binary",
        (_fa_high_lbl, _fa_low_lbl)),
]

_rows = []
for pred_name, col, kind, levels in PRED_SPECS:
    if kind == "binary":
        cat_a, cat_b = levels
        nA, pA, nB, pB, p_eff = _two_cat_row(
            _uni, col, cat_a, cat_b, "Effective", LBL_EFFECTIVE)
        n_iA, pi_A, n_iB, pi_B, p_int = _two_cat_row(
            _uni, col, cat_a, cat_b, "Condition", "Intact")
        _rows.append({
            "Predictor": pred_name,
            "Level": cat_a,
            "n": nA,
            "Test": "Fisher (2-sided)",
            "% veg response": f"{pA*100:.1f}",
            "Veg p (adj.)":   _fmt_p(p_eff),
            "Veg sig":        _stars(p_eff),
            "% intact":       f"{pi_A*100:.1f}",
            "Cond p (adj.)":  _fmt_p(p_int),
            "Cond sig":       _stars(p_int),
        })
        _rows.append({
            "Predictor": "",
            "Level": cat_b,
            "n": nB,
            "Test": "",
            "% veg response": f"{pB*100:.1f}",
            "Veg p (adj.)":   "",
            "Veg sig":        "",
            "% intact":       f"{pi_B*100:.1f}",
            "Cond p (adj.)":  "",
            "Cond sig":       "",
        })
    else:  # multi-cat
        rows_eff, padj_eff = _multi_cat_summary(
            _uni, col, levels, "Effective", LBL_EFFECTIVE)
        rows_int, padj_int = _multi_cat_summary(
            _uni, col, levels, "Condition", "Intact")
        cats_eff = [r[0] for r in rows_eff]
        cats_int = [r[0] for r in rows_int]
        # Both outcome maps should yield the same level order (same group column);
        # use the union, preserving levels order from cat_order.
        all_levels = [c for c in levels if c in cats_eff or c in cats_int]
        first = True
        for lvl in all_levels:
            try:
                ie = cats_eff.index(lvl)
                n_e, pe = rows_eff[ie][1], rows_eff[ie][2]
                pairs_e = _sig_pairs_string(ie, cats_eff, padj_eff)
            except ValueError:
                n_e, pe, pairs_e = np.nan, np.nan, ""
            try:
                ii = cats_int.index(lvl)
                n_i, pi = rows_int[ii][1], rows_int[ii][2]
                pairs_i = _sig_pairs_string(ii, cats_int, padj_int)
            except ValueError:
                n_i, pi, pairs_i = np.nan, np.nan, ""
            n_use = n_e if not pd.isna(n_e) else n_i
            _rows.append({
                "Predictor":      pred_name if first else "",
                "Level":          lvl,
                "n":              int(n_use) if not pd.isna(n_use) else "",
                "Test":           ("Pairwise Fisher (BH-FDR)" if first else ""),
                "% veg response": f"{pe*100:.1f}" if not pd.isna(pe) else "",
                "Veg p (adj.)":   pairs_e,
                "Veg sig":        "",
                "% intact":       f"{pi*100:.1f}" if not pd.isna(pi) else "",
                "Cond p (adj.)":  pairs_i,
                "Cond sig":       "",
            })
            first = False

univariate_outcomes_table = pd.DataFrame(_rows)
display(univariate_outcomes_table)

# Save CSV
_out_csv = Path("../data/summary/univariate_outcomes_table.csv")
_out_csv.parent.mkdir(parents=True, exist_ok=True)
univariate_outcomes_table.to_csv(_out_csv, index=False)
print(f"\nSaved: {_out_csv}")
```

```python
# ── Covariate association matrix (mixed numeric + categorical predictors) ──
# Spearman's rho is only meaningful for ordered (numeric/ordinal) variables.
# For nominal categoricals (incl. binary) it collapses to the phi coefficient
# and ignores any non-monotonic structure, so we use type-appropriate measures
# that are all symmetric and bounded on [0, 1]:
#   numeric  vs numeric   ->  |Spearman rho|              (monotonic rank association)
#   category vs category  ->  Cramer's V (Bergsma-Wicher) (chi-square based, bias-corrected)
#   numeric  vs category  ->  correlation ratio eta       (variance explained)
from pathlib import Path
import shutil as _shutil
from scipy.stats import spearmanr as _spearmanr, chi2_contingency as _chi2_contingency

_FS = FS

# Predictor set mirrors the PCA cell: 3 numeric + 3 categorical, plus 2 binary outcomes
_assoc_num = ["slope_200", "FA_30_max", "Shape_Leng"]
_assoc_cat = ["Landform", "Texture", "Soil_Development"]
_assoc_out = ["Condition", "Effective"]   # binary outcomes (Intact/Degraded, veg response)
_assoc_cols = _assoc_num + _assoc_cat + _assoc_out

_df_assoc = data.dropna(subset=_assoc_cols).copy()

# Keep only the 3 most frequent soil textures (consistent with the PCA figure)
_top3_tex = _df_assoc["Texture"].value_counts().nlargest(3).index.tolist()
_df_assoc = _df_assoc[_df_assoc["Texture"].isin(_top3_tex)].copy()
_n_assoc = len(_df_assoc)


def _cramers_v_corrected(a, b):
    """Bias-corrected Cramer's V (Bergsma-Wicher 2013) for two categorical series, [0, 1]."""
    _ct = pd.crosstab(a, b)
    if _ct.shape[0] < 2 or _ct.shape[1] < 2:
        return 0.0
    _chi2 = _chi2_contingency(_ct)[0]
    _n = _ct.values.sum()
    _phi2 = _chi2 / _n
    _r, _k = _ct.shape
    _phi2_c = max(0.0, _phi2 - (_r - 1) * (_k - 1) / (_n - 1))
    _r_c = _r - (_r - 1) ** 2 / (_n - 1)
    _k_c = _k - (_k - 1) ** 2 / (_n - 1)
    _denom = min(_r_c - 1, _k_c - 1)
    return float(np.sqrt(_phi2_c / _denom)) if _denom > 0 else 0.0


def _correlation_ratio(cats, values):
    """Correlation ratio eta between a categorical and a numeric series, [0, 1]."""
    _cats = pd.Series(cats).astype("category")
    _vals = pd.Series(values).astype(float).values
    _grand = _vals.mean()
    _ss_total = np.sum((_vals - _grand) ** 2)
    if _ss_total == 0:
        return 0.0
    _ss_between = 0.0
    for _code in _cats.cat.categories:
        _grp = _vals[(_cats == _code).values]
        if len(_grp):
            _ss_between += len(_grp) * (_grp.mean() - _grand) ** 2
    return float(np.sqrt(_ss_between / _ss_total))


# ── Build symmetric association matrix ────────────────────────────────────
_assoc = pd.DataFrame(np.eye(len(_assoc_cols)),
                      index=_assoc_cols, columns=_assoc_cols, dtype=float)
for _i, _ci in enumerate(_assoc_cols):
    for _cj in _assoc_cols[_i + 1:]:
        _i_num, _j_num = _ci in _assoc_num, _cj in _assoc_num
        if _i_num and _j_num:                       # numeric vs numeric
            _v = abs(_spearmanr(_df_assoc[_ci], _df_assoc[_cj]).correlation)
        elif (not _i_num) and (not _j_num):         # categorical vs categorical
            _v = _cramers_v_corrected(_df_assoc[_ci], _df_assoc[_cj])
        else:                                       # mixed
            _cat_col, _num_col = (_ci, _cj) if not _i_num else (_cj, _ci)
            _v = _correlation_ratio(_df_assoc[_cat_col], _df_assoc[_num_col])
        _assoc.loc[_ci, _cj] = _assoc.loc[_cj, _ci] = _v

# Match manuscript terminology in the figure labels
_assoc_label_overrides = {"Condition": "Structural condition", "Effective": "Vegetation response"}
_assoc_labels = [_assoc_label_overrides.get(_c, PRETTY_LABELS.get(_c, _c)) for _c in _assoc_cols]
_assoc_disp = _assoc.copy()
_assoc_disp.index = _assoc_labels
_assoc_disp.columns = _assoc_labels

# ── Plot ──────────────────────────────────────────────────────────────────
_fig_as, _ax_as = plt.subplots(figsize=(8.6, 7.2))
# Mask the upper triangle and the all-ones diagonal (k=0); they carry no information
_mask = np.triu(np.ones_like(_assoc_disp, dtype=bool), k=0)
sns.heatmap(_assoc_disp, mask=_mask, cmap="Blues", vmin=0, vmax=1,
            annot=True, fmt=".2f", annot_kws={"fontsize": _FS},
            linewidths=0.6, linecolor="white", square=True,
            cbar_kws={"label": "Association strength", "shrink": 0.8},
            ax=_ax_as)
_ax_as.tick_params(labelsize=_FS)
_cbar = _ax_as.collections[0].colorbar
_cbar.ax.tick_params(labelsize=_FS)
_cbar.ax.yaxis.label.set_size(_FS)
plt.setp(_ax_as.get_xticklabels(), rotation=35, ha="right")
plt.setp(_ax_as.get_yticklabels(), rotation=0)
plt.tight_layout()

# ── Save: registry copy + manuscript copy ─────────────────────────────────
_out_as = Path("../figures/outcomes/SI_predictor_association.png")
_out_as.parent.mkdir(parents=True, exist_ok=True)
_fig_as.savefig(_out_as, dpi=300, bbox_inches="tight")

# Copy into the manuscript Figures folder as Figure 3
_manu_as = Path("../draft/local/Figures/fig3_predictor_association.png")
if _manu_as.parent.exists():
    _shutil.copyfile(_out_as, _manu_as)
    print(f"Copied -> {_manu_as}")

# Strongest off-diagonal pair for the dynamic description
_off = _assoc.where(~np.eye(len(_assoc_cols), dtype=bool))
_imax, _jmax = np.unravel_index(np.nanargmax(_off.values), _off.shape)
_max_pair = (PRETTY_LABELS.get(_assoc_cols[_imax], _assoc_cols[_imax]),
             PRETTY_LABELS.get(_assoc_cols[_jmax], _assoc_cols[_jmax]))
_max_val = _off.values[_imax, _jmax]

register_outcomes_figure(
    "SI_PREDICTOR_ASSOCIATION",
    label="SI_assoc",
    file_name=_out_as.name,
    stats_text=(
        f"Pairwise association among {len(_assoc_cols)} variables "
        f"(3 numeric + 3 categorical predictors + 2 binary outcomes \u2014 structural condition and "
        f"vegetation response; top 3 textures only, n = {_n_assoc}). "
        f"Numeric\u2013numeric pairs use |Spearman \u03c1|, categorical\u2013categorical pairs use "
        f"the Bergsma-Wicher bias-corrected Cramer's V, and mixed pairs use the correlation ratio \u03b7; "
        f"all bounded on [0, 1]. Strongest pair: {_max_pair[0]} \u2013 {_max_pair[1]} ({_max_val:.2f})."
    ),
    interpretation_text=(
        "Spearman's \u03c1 is appropriate only for ordered numeric pairs; for nominal categoricals it "
        "reduces to the phi coefficient and discards non-monotonic structure, so Cramer's V and the "
        "correlation ratio are used instead. "
        f"The largest association ({_max_pair[0]}\u2013{_max_pair[1]}, {_max_val:.2f}) indicates predictors are "
        "only moderately collinear, while structural condition and vegetation response show only weak "
        "association with the predictors and with each other, supporting their joint use in the PCA "
        "and predictor-ranking analyses."
    ),
)

plt.show()
```

```python
# ── Failure rate per unit length (continuous) ──────────────────────────────
import statsmodels.api as sm
from scipy.stats import spearmanr, mannwhitneyu

_FS = FS

# Working copy — all berms with length and condition data
_df_len = data.dropna(subset=['Shape_Leng', 'Condition']).copy()
_df_len['_failed'] = (_df_len['Condition'] == 'Degraded').astype(int)
_n_total = len(_df_len)

# ── 1. Logistic regression: failure ~ Shape_Leng ─────────────────────────
_X_len = sm.add_constant(_df_len['Shape_Leng'].values)
_logit_len = sm.Logit(_df_len['_failed'].values, _X_len).fit(disp=0)

print('─── Logistic regression: Degraded ~ Shape_Leng ───')
print(f'  β(Shape_Leng) = {_logit_len.params[1]:.5f}  (p = {_logit_len.pvalues[1]:.4f})')
print(f'  Pseudo-R²     = {_logit_len.prsquared:.4f}')
print(f'  n = {_logit_len.nobs:.0f}')

# Odds ratio per 10 m increase
_or_10m = np.exp(_logit_len.params[1] * 10)
print(f'  OR per 10 m increase: {_or_10m:.3f}')

# ── 2. Spearman correlation (point-biserial equivalent for ordinal) ───────
_rho, _p_spear = spearmanr(_df_len['Shape_Leng'], _df_len['_failed'])
print(f'\n─── Spearman: Shape_Leng vs Degraded ───')
print(f'  ρ = {_rho:.3f}, p = {_p_spear:.4f}')

# ── 3. By Fail_Type — length distributions and comparisons ───────────────
_df_failed = _df_len[_df_len['_failed'] == 1].copy()
_n_failed = len(_df_failed)

if 'Fail_Type' in _df_failed.columns:
    print(f'\n─── Length by Fail_Type (failed berms, n = {_n_failed}) ───')
    for _ft in ['Breach', 'Flank']:
        _sub = _df_failed[_df_failed['Fail_Type'] == _ft]
        print(f'  {_ft} (n={len(_sub)}): median = {_sub["Shape_Leng"].median():.1f} m, '
              f'mean = {_sub["Shape_Leng"].mean():.1f} m')

    # Compare raw lengths between breach and flank
    _U_len, _p_len = mannwhitneyu(
        _df_failed.loc[_df_failed['Fail_Type'] == 'Breach', 'Shape_Leng'],
        _df_failed.loc[_df_failed['Fail_Type'] == 'Flank',  'Shape_Leng'],
        alternative='two-sided')
    print(f'  Breach vs Flank length: U = {_U_len:.0f}, p = {_p_len:.4f}')

    # Intact vs all failed
    _intact_len = _df_len.loc[_df_len['_failed'] == 0, 'Shape_Leng']
    _failed_len = _df_len.loc[_df_len['_failed'] == 1, 'Shape_Leng']
    _U_if, _p_if = mannwhitneyu(_intact_len, _failed_len, alternative='two-sided')
    print(f'\n  Intact (n={len(_intact_len)}): median = {_intact_len.median():.1f} m')
    print(f'  Failed (n={len(_failed_len)}): median = {_failed_len.median():.1f} m')
    print(f'  Intact vs Failed length: U = {_U_if:.0f}, p = {_p_if:.4f}')
```

### Interaction between intactness and effectiveness

```python

# ── Fisher's exact test: effectiveness ~ Intact/Degraded, repeated across grouping variables ──
# For each grouping variable (Landform, ParentMaterial, Soil_Development, Texture),
# test whether the proportion of effective berms differs between Intact and Degraded.

from scipy.stats import fisher_exact

MIN_N_PER_CATEGORY = 10   # minimum berms per Intact/Degraded group to run Fisher test

def fisher_by_group(data, group_col, min_n=10):
    """
    For each level of group_col, test whether effectiveness differs
    between Intact (True) and Degraded (False) berms using Fisher's exact test.
    Only tests groups where both Intact and Degraded have >= min_n samples.
    Returns a DataFrame with one row per group level.
    """
    sub = data.dropna(subset=[group_col, "Intact", "Effective"]).copy()
    rows = []
    for grp_val, g in sub.groupby(group_col):
        ct = pd.crosstab(g["Intact"], g["Effective"].map({LBL_EFFECTIVE: True, LBL_INEFFECTIVE: False})).reindex(
            index=[False, True], columns=[False, True], fill_value=0
        )
        n_deg = int(ct.loc[False].sum())
        n_int = int(ct.loc[True].sum())
        p_deg = ct.loc[False, True] / n_deg if n_deg > 0 else np.nan
        p_int = ct.loc[True, True] / n_int if n_int > 0 else np.nan

        if min(n_deg, n_int) >= min_n:
            table = [
                [ct.loc[False, False], ct.loc[False, True]],
                [ct.loc[True,  False], ct.loc[True,  True]],
            ]
            _, p_val = fisher_exact(table, alternative="two-sided")
        else:
            p_val = np.nan

        rows.append({
            "Variable": group_col,
            "Group": grp_val,
            "n_degraded": n_deg,
            "n_intact": n_int,
            "p_eff_degraded": round(p_deg, 3) if pd.notna(p_deg) else np.nan,
            "p_eff_intact":   round(p_int, 3) if pd.notna(p_int) else np.nan,
            "diff (intact-degraded)": round(p_int - p_deg, 3) if pd.notna(p_int) and pd.notna(p_deg) else np.nan,
            "p_value": round(p_val, 4) if pd.notna(p_val) else np.nan,
            "significant": (p_val < 0.05) if pd.notna(p_val) else False,
        })
    return pd.DataFrame(rows).sort_values("p_value").reset_index(drop=True)


# Run across all grouping variables
GROUP_VARS = ["Landform", "ParentMaterial", "Soil_Development", "Texture", "Slope_Class", "Berm_Length_Class"]
all_results = pd.concat(
    [fisher_by_group(data, v, min_n=MIN_N_PER_CATEGORY) for v in GROUP_VARS],
    ignore_index=True,
)

# ── Per-variable tables ──
for var in GROUP_VARS:
    sub_res = all_results[all_results["Variable"] == var].copy()
    print(f"\n{'═'*60}")
    print(f"  {var}  (Fisher's exact, two-sided, min n={MIN_N_PER_CATEGORY} per group)")
    print(f"{'═'*60}")
    display(sub_res.drop(columns="Variable").reset_index(drop=True))

# ── Combined summary: only significant results ──
print("\n\n" + "═"*60)
print("  SUMMARY: Significant differences (p < 0.05)")
print("═"*60)
sig = all_results[all_results["significant"]].reset_index(drop=True)
if len(sig):
    display(sig)
else:
    print("No significant differences found.")

# ── Count of significant results per variable ──
print("\nSignificant pairs per variable:")
print(all_results.groupby("Variable")["significant"].sum().to_string())
```

```python
# ── Fig 3: Intact × Vegetation response co-occurrence  +  vegetation response
#           by soil texture & condition ──────────────────────────────────────────

import matplotlib.gridspec as gridspec
from scipy.stats import pearsonr, fisher_exact as _fisher_exact

_FS = FS  # unified font size for the figure

# ── PANEL A data: co-occurrence heatmap ──────────────────────────────────────
_valid = data[["Intact", "Effective"]].dropna().copy()
_valid["_int_bin"] = _valid["Intact"].astype(bool).astype(int)
_valid["_eff_bin"] = (_valid["Effective"] == LBL_EFFECTIVE).astype(int)

_ct_cross = (
    pd.crosstab(_valid["_int_bin"], _valid["_eff_bin"],
                rownames=["Intactness"], colnames=["Vegetation response"])
    .reindex(index=[0, 1], columns=[0, 1], fill_value=0)
)
_ct_pct = (_ct_cross / _ct_cross.values.sum() * 100).round(1)
_r_phi, _p_phi = pearsonr(_valid["_int_bin"], _valid["_eff_bin"])

# ── PANEL B data: proportion with veg response by texture & condition (top 3) ─
_sub = data.dropna(subset=["Texture", "Effective"]).copy()
_sub["_cond"] = _sub["Condition"] if "Condition" in _sub.columns else _sub["Intact"].map({True: "Intact", False: "Degraded", 1: "Intact", 0: "Degraded"})
_sub["_eff"]  = (_sub["Effective"] == LBL_EFFECTIVE).astype(int)

# Top 3 textures by total count
_top3 = _sub["Texture"].value_counts().head(3).index.tolist()
_sub  = _sub[_sub["Texture"].isin(_top3)]

_rows_b = []
for (tex, cond), g in _sub.groupby(["Texture", "_cond"]):
    _rows_b.append({"Texture": tex, "Group": cond, "count": len(g),
                    "eff_count": int(g["_eff"].sum()), "mean": g["_eff"].mean()})
_plot_df = pd.DataFrame(_rows_b)

# Fisher's exact per texture
_pvals = {}
for tex in _plot_df["Texture"].unique():
    _gt = _sub[_sub["Texture"] == tex]
    _ct_tex = pd.crosstab(_gt["_cond"], _gt["_eff"]).reindex(
        index=["Degraded", "Intact"], columns=[0, 1], fill_value=0)
    if _ct_tex.shape == (2, 2) and _ct_tex.sum().sum() > 0:
        _, _pv = _fisher_exact(_ct_tex.values, alternative="two-sided")
        _pvals[tex] = _pv
    else:
        _pvals[tex] = np.nan
_plot_df["p_value"] = _plot_df["Texture"].map(_pvals)
_plot_df["sig"]     = _plot_df["p_value"] < 0.05

_known_tex = set(texture_order)
_tex_order = [t for t in texture_order if t in _top3] + \
             [t for t in _top3 if t not in set(texture_order)]

# ── Figure layout ─────────────────────────────────────────────────────────────
fig = plt.figure(figsize=(15, 5.5))
gs  = gridspec.GridSpec(1, 2, width_ratios=[1, 1], wspace=0.30)
ax1 = fig.add_subplot(gs[0])
ax2 = fig.add_subplot(gs[1])

# ── PANEL A: 2×2 heatmap ─────────────────────────────────────────────────────
_arr     = _ct_cross.values
_arr_pct = _ct_pct.values
ax1.imshow(_arr, cmap=plt.cm.Blues, vmin=0, vmax=_arr.max() * 1.35, aspect="auto")

_row_labels = ["Degraded", "Intact"]
_col_labels = ["No response", "Vegetation\nresponse"]

for i in range(2):
    for j in range(2):
        _tc = "white" if _arr[i, j] > _arr.max() * 0.55 else "black"
        ax1.text(j, i, f"{_arr[i, j]}\n({_arr_pct[i, j]:.1f}%)",
                 ha="center", va="center", fontsize=_FS,
                 fontweight="bold", color=_tc)

# Red border on off-diagonal (discordant) cells
for i, j in [(0, 1), (1, 0)]:
    ax1.add_patch(plt.Rectangle(
        (j - 0.48, i - 0.48), 0.96, 0.96,
        fill=False, edgecolor="tomato", linewidth=2.2, zorder=5))

ax1.set_xticks([0, 1]); ax1.set_xticklabels(_col_labels, fontsize=_FS)
ax1.set_yticks([0, 1]); ax1.set_yticklabels(_row_labels, fontsize=_FS)
ax1.set_xlabel("Vegetation response", fontsize=_FS)
ax1.set_ylabel("Condition", fontsize=_FS)

_p_str = f"p = {_p_phi:.3f}" if _p_phi >= 0.001 else "p < 0.001"
_assoc = "no significant association" if _p_phi > 0.05 else "significant"
ax1.set_title(f"A  |  Intact × Vegetation response co-occurrence\n"
              f"φ = {_r_phi:.2f}  ({_p_str};  {_assoc})",
              fontsize=_FS, fontweight="bold", loc="left", pad=14)

# ── PANEL B: grouped horizontal barplot ───────────────────────────────────────
_n_tex  = len(_tex_order)
_bar_w  = 0.35
_y_pos  = np.arange(_n_tex)
_clr    = {"Intact": INTACT_COL, "Degraded": DEGRADED_COL}

for idx, (grp, color) in enumerate(_clr.items()):
    _grp_df = _plot_df[_plot_df["Group"] == grp].set_index("Texture")
    _vals   = [_grp_df.loc[t, "mean"] if t in _grp_df.index else 0 for t in _tex_order]
    _cnts   = [int(_grp_df.loc[t, "eff_count"]) if t in _grp_df.index else 0 for t in _tex_order]
    _offset = -_bar_w / 2 if idx == 0 else _bar_w / 2
    _bars = ax2.barh(_y_pos + _offset, _vals, _bar_w, label=grp, color=color,
                     edgecolor="black", linewidth=0.7)
    # Label each bar with its count
    for _bar, _cnt in zip(_bars, _cnts):
        _bw = _bar.get_width()
        ax2.text(_bw + 0.01, _bar.get_y() + _bar.get_height() / 2,
                 f"n={_cnt}", va="center", fontsize=_FS, color="#333333")

# Significance annotations
for j, tex in enumerate(_tex_order):
    _row_i = _plot_df[(_plot_df["Texture"] == tex) & (_plot_df["Group"] == "Intact")]
    _row_d = _plot_df[(_plot_df["Texture"] == tex) & (_plot_df["Group"] == "Degraded")]
    if _row_i.empty or _row_d.empty:
        continue
    if _row_i["sig"].values[0]:
        _p = _row_i["p_value"].values[0]
        _top = max(_row_i["mean"].values[0], _row_d["mean"].values[0])
        ax2.text(_top + 0.12, j, f"p={_p:.3f}*",
                 va="center", fontsize=_FS, color="#333333")

ax2.set_yticks(_y_pos)
ax2.set_yticklabels(_tex_order, fontsize=_FS)
ax2.set_xlabel("Proportion with vegetation response", fontsize=_FS)
ax2.set_xlim(0, 1.15)
ax2.invert_yaxis()
ax2.axvline(0.5, color="#cccccc", linewidth=0.8, linestyle="--")
ax2.legend(title="Condition", fontsize=_FS, title_fontsize=_FS)
ax2.spines[["top", "right"]].set_visible(False)
ax2.tick_params(axis="x", labelsize=_FS)
ax2.set_title("B  |  Vegetation response by soil texture & condition\n"            ,
              fontsize=_FS, fontweight="bold", loc="left", pad=14)


# ── Save and register ─────────────────────────────────────────────────────────
_name = "fig8_veg_response_by_condition_texture.png"
_out  = Path(f"../figures/outcomes/{_name}")
_out.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(_out, dpi=300, bbox_inches="tight")
print(f"Saved → {_out}")

register_outcomes_figure(
    "FIG_8",
    label="fig8",
    file_name=_name,
    stats_text=(
        f"Panel A: φ = {_r_phi:.2f}, p = {_p_phi:.3f}. "
        "Panel B: Fisher's exact tests per texture (top 3 by sample size); proportions with count labels."
    ),
    interpretation_text=(
        "Structural condition and vegetation response are statistically independent (φ ≈ 0). "
        "Sandy loam shows the only significant difference in vegetation response between "
        "intact and degraded berms."
    ),
)
plt.show()
```
